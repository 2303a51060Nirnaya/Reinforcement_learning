```python
import random

machines = ['A', 'B', 'C']
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]
epsilon = 0.2
N_play = 10

for play in range(1, N_play + 1):
    print("\nPlay", play)

    r = random.random()

    # ε-Greedy decision
    if r < epsilon:
        i = random.randint(0, 2)
        print("Exploration")
    else:
        i = Q.index(max(Q))
        print("Exploitation")

    print("Selected Machine:", machines[i])

    # Reward collection
    reward = float(input("Enter reward: "))

    # Update N
    N[i] += 1

    # Update R
    R[i] += reward

    # Incremental sample-average update
    Q[i] = Q[i] + (reward - Q[i]) / N[i]

    print("N =", N)
    print("R =", R)
    print("Q =", Q)

print("\nFinal Results")
print("Machine:", machines)
print("N =", N)
print("R =", R)
print("Q =", Q)
```
