```python
n = int(input("Enter number of machines: "))

machines = []

for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))
    q = sum(rewards) / len(rewards)
    machines.append((name, rewards, q))

print("\nResults:")

for name, rewards, q in machines:
    print(f"\nMachine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q:.2f}")

selected = max(machines, key=lambda x: x[2])

print(f"\nGreedy Selected Machine: {selected[0]}")
```

```python
n = int(input("Enter number of machines: "))

names = []
rewards = []

for i in range(n):
    names.append(input("Enter machine name: "))
    rewards.append(list(map(int, input("Enter rewards: ").split())))

Q = [0.0] * n
N = [0] * n

for i in range(n):
    print(f"\nMachine {names[i]}")

    for R in rewards[i]:
        N[i] += 1
        Q[i] = Q[i] + (R - Q[i]) / N[i]

        print(f"Reward = {R}, N = {N[i]}, Q({names[i]}) = {Q[i]:.2f}")

        selected = max(range(n), key=lambda x: Q[x])
        print(f"Greedy Selected: {names[selected]}")

print("\nFinal Q Values:")
for i in range(n):
    print(f"Q({names[i]}) = {Q[i]:.2f}")

selected = max(range(n), key=lambda x: Q[x])
print(f"\nFinal Greedy Selected Machine: {names[selected]}")
```
